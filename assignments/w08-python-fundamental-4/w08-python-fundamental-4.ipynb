## Bài 1.

```python
def tim_so(a, b):
    ds = []
    for i in range(a, b + 1):
        if i % 7 == 0 and i % 5 != 0:
            ds.append(str(i))
    return ds
print(", ".join(tim_so(2000, 3200)))
```

## Bài 2.

```python
def giai_thua(n):
    kq = 1
    for i in range(1, n + 1):
        kq = kq * i
    return kq
n = int(input("Nhập số cần tính giai thừa: "))
print("Giai thừa của", n, "là:", giai_thua(n))
```

## Bài 3.

```python
def tao_dictionary(n):
    d = {}
    for i in range(1, n + 1):
        d[i] = i * i
    return d
n = int(input("Nhập số nguyên n: "))
print("Với n =", n, "thì dictionary là:", tao_dictionary(n))
```

## Bài 9.

```python
import datetime

def tinh_tuoi(nam, thang, ngay):
    hom_nay = datetime.date.today()
    tuoi = hom_nay.year - nam
    if (hom_nay.month, hom_nay.day) < (thang, ngay):
        tuoi = tuoi - 1
    return tuoi
s = input("Nhập ngày tháng năm sinh: ")
nam, thang, ngay = s.split("/")
print("Ngày sinh vừa nhập:", s)
print("Số tuổi là:", tinh_tuoi(int(nam), int(thang), int(ngay)))
```

## Bài 10.

```python
def tinh_luong(so_gio, thu_lao):
    if so_gio <= 44:
        return so_gio * thu_lao
    return 44 * thu_lao + (so_gio - 44) * thu_lao * 1.5
so_gio = float(input("Nhập số giờ làm mỗi tuần: "))
thu_lao = float(input("Nhập thù lao mỗi giờ làm tiêu chuẩn: "))
print("Số giờ làm mỗi tuần:", so_gio)
print("Thù lao mỗi giờ làm tiêu chuẩn:", thu_lao)
print("Số tiền thực lĩnh là:", tinh_luong(so_gio, thu_lao))
```

## Bài 11.

```python
import math

def tinh_q(d):
    c = 50
    h = 30
    return round(math.sqrt(2 * c * d / h))
s = input("Nhập dãy giá trị D: ")
ket_qua = []
for d in s.split(","):
    ket_qua.append(str(tinh_q(int(d))))
print("Các giá trị D vừa nhập:", s)
print("Các giá trị Q là:", ",".join(ket_qua))
```

## Bài 12.

```python
def tao_mang(x, y):
    mang = []
    for i in range(x):
        hang = []
        for j in range(y):
            hang.append(i * j)
        mang.append(hang)
    return mang
s = input("Nhập X, Y: ")
x, y = s.split(",")
print("X =", x, "; Y =", y)
print("Mảng 2 chiều:", tao_mang(int(x), int(y)))
```

## Bài 13.

```python
def sap_xep_tu(s):
    cac_tu = []
    for tu in s.split(","):
        if tu != "":
            cac_tu.append(tu)
    cac_tu.sort()
    return cac_tu
s = input("Nhập các từ: ")
print("Các từ vừa nhập:", s)
print("Các từ sau khi sắp xếp:", ", ".join(sap_xep_tu(s)))
```

## Bài 14.

```python
def chuyen_hoa(cac_dong):
    ket_qua = []
    for dong in cac_dong:
        ket_qua.append(dong.upper())
    return ket_qua
cac_dong = []
while True:
    dong = input()
    if dong == "":
        break
    cac_dong.append(dong)
print("Các dòng vừa nhập:")
for dong in cac_dong:
    print(dong)
print("Các dòng sau khi chuyển thành chữ hoa:")
for dong in chuyen_hoa(cac_dong):
    print(dong)
```

```python
def chia_het_cho_5(s):
    ds = []
    for so in s.split(","):
        if int(so, 2) % 5 == 0:
            ds.append(so)
    return ds
s = input("Nhập các số nhị phân 4 chữ số: ")
print("Các số nhị phân vừa nhập:", s)
print("Các số chia hết cho 5:", ", ".join(chia_het_cho_5(s)))
```

## Bài 17.

```python
def toan_chu_so_chan(n):
    while n > 0:
        if (n % 10) % 2 != 0:
            return False
        n = n // 10
    return True
def tim_so(a, b):
    ds = []
    for i in range(a, b + 1):
        if toan_chu_so_chan(i):
            ds.append(str(i))
    return ds
print(", ".join(tim_so(1000, 3000)))
```

## Bài 18.

```python
def dem_chu_cai_va_chu_so(s):
    chu_cai = 0
    chu_so = 0
    for ch in s:
        if ch.isalpha():
            chu_cai = chu_cai + 1
        elif ch.isdigit():
            chu_so = chu_so + 1
    return chu_cai, chu_so
s = input("Nhập một câu: ")
chu_cai, chu_so = dem_chu_cai_va_chu_so(s)
print("Câu vừa nhập:", s)
print("Số chữ cái là:", chu_cai)
print("Số chữ số là:", chu_so)
```

## Bài 22.

```python
def tinh_so_tien(nhat_ky):
    tien = 0
    for dong in nhat_ky:
        loai, so = dong.split()
        if loai == "D":
            tien = tien + int(so)
        elif loai == "W":
            tien = tien - int(so)
    return tien
nhat_ky = []
while True:
    dong = input()
    if dong == "":
        break
    nhat_ky.append(dong)
print("Nhật ký giao dịch vừa nhập:")
for dong in nhat_ky:
    print(dong)
print("Số tiền thực của tài khoản là:", tinh_so_tien(nhat_ky))
```

## Bài 23.

```python
def kiem_tra_doi_xung(A):
    return A == A[::-1]

def sx_doi_xung(A):
    n = len(A)
    for i in range(n // 2):
        if A[i] != A[n - 1 - i]:
            A[n - 1 - i] = A[i]
    return A
n = int(input("Nhập số n: "))
A = []
for i in range(n):
    A.append(int(input("Nhập phần tử thứ " + str(i + 1) + ": ")))
print("Dãy số vừa nhập:", end=" ")
for x in A:
    print(x, end=" ")
print()
if kiem_tra_doi_xung(A):
    print("Dãy số đã đối xứng")
else:
    print("Dãy số chưa đối xứng")
    A = sx_doi_xung(A)
    print("Dãy số sau khi biến đổi thành đối xứng:", end=" ")
    for x in A:
        print(x, end=" ")
    print()
```
