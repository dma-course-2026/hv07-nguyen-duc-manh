```python
def tinh_tien_dien(kw):
    if kw <= 60:
        tien = kw * 5
    elif kw <= 160:
        tien = 60 * 5 + (kw - 60) * 8
    else:
        tien = 60 * 5 + 100 * 8 + (kw - 160) * 10
    return tien
ho_ten = input("Họ và tên: ")
chiso1 = int(input("Nhập chỉ số tháng trước: "))
chiso2 = int(input("Nhập chỉ số tháng này: "))
print("Họ và tên:", ho_ten)
print("Tiền phải trả là", float(tinh_tien_dien(chiso2 - chiso1)))
```

## Bài 20. Phương trình bậc 2

```python
import math

def giai_bac_1(a, b):
    if a != 0:
        print("Phương trình có một nghiệm! x =", -b / a)
    elif b == 0:
        print("Phương trình có vô số nghiệm")
    else:
        print("Phương trình vô nghiệm")
def giai_bac_2(a, b, c):
    if a == 0:
        giai_bac_1(b, c)   
        return             
    delta = b * b - 4 * a * c
    if delta < 0:
        print("Phương trình vô nghiệm")
    elif delta == 0:
        print("Phương trình có nghiệm kép: x =", -b / (2 * a))
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print("Phương trình có 2 nghiệm phân biệt:")
        print("x1 =", x1)
        print("x2 =", x2)
a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
c = float(input("Nhập c: "))
giai_bac_2(a, b, c)
```

## Bài 33. Điểm thuộc đường tròn

```python
import math

def khoang_cach(x1, y1, x2, y2):
    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
def vi_tri_diem(R, x0, y0, xa, ya):
    d = khoang_cach(x0, y0, xa, ya)
    if abs(d - R) < 0.000001:      
        print("Điểm A nằm trên đường tròn")
    elif d < R:
        print("Điểm A nằm trong đường tròn")
    else:
        print("Điểm A nằm ngoài đường tròn")
R = float(input("Nhập bán kính R: "))
print("Nhập tọa độ tâm đường tròn:")
x0 = float(input("x0: "))
y0 = float(input("y0: "))
print("Nhập tọa độ điểm A:")
xa = float(input("xa: "))
ya = float(input("ya: "))
vi_tri_diem(R, x0, y0, xa, ya)
```

## Bài 43. Đếm số ước của n

```python
def dem_uoc(n):
    S = 0
    i = 1
    while i * i <= n:              
        if n % i == 0:
            if i * i == n:
                S = S + 1      
            else:
                S = S + 2      
        i = i + 1
    return S


n = int(input("Nhập n: "))
print(n, "có", dem_uoc(n), "ước")
```

```python
def tong_uoc_thuc_su(n):
    S = 0
    for i in range(1, n):
        if n % i == 0:
            S = S + i
    return S
def hoan_hao(n):
    return n > 1 and tong_uoc_thuc_su(n) == n
n = int(input("Nhập số n: "))
for i in range(1, n):
    if hoan_hao(i):
        print(i)
```

## Bài 48. Số chính phương

```python
import math

def so_chinh_phuong(n):
    if n <= 0:
        return False
    m = int(math.sqrt(n))
    return m * m == n or (m + 1) * (m + 1) == n
n = int(input("Nhập n: "))
if so_chinh_phuong(n):
    print(n, "là số chính phương")
else:
    print(n, "không là số chính phương")
```

## Bài 49. Số 3 chữ số abc = a³ + b³ + c³ (số Armstrong)

```python
def so_armstrong(a, b, c):
    return 100 * a + 10 * b + c == a ** 3 + b ** 3 + c ** 3
for a in range(1, 10):          
    for b in range(10):
        for c in range(10):
            if so_armstrong(a, b, c):
                print(a, b, c)
```

## Bài 50. Số 3 chữ số thoả a + b + c = a·b·c

```python
def tong_bang_tich(a, b, c):
    return a + b + c == a * b * c
for a in range(1, 10):
    for b in range(10):
        for c in range(10):
            if tong_bang_tich(a, b, c):
                print(a, b, c)
```
