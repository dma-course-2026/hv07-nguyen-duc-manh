## Bài 20. Phương trình bậc 2

```python
import math

a = float(input("nhập a: "))
b = float(input("nhập b: "))
c = float(input("nhập c: "))

if a == 0:  # suy biến về bậc 1
    if b != 0:
        print("Phương trình có 1 nghiệm: x =", -c / b)
    elif c == 0:
        print("Phương trình có vô số nghiệm")
    else:
        print("Phương trình vô nghiệm")
else:
    delta = b * b - 4 * a * c
    if delta < 0:
        print("Phương trình vô nghiệm")
    elif delta == 0:
        print("Phương trình có nghiệm kép: x =", -b / (2 * a))
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print("Phương trình có 2 nghiệm phân biệt:")
        print("x1 =", x1)
        print("x2 =", x2)
```

## Bài 33. Điểm thuộc đường tròn

```python
import math

R = float(input("Nhập bán kính R: "))
print("Nhập tọa độ tâm đường tròn:")
x0 = float(input("x0: "))
y0 = float(input("y0: "))
print("Nhập tọa độ điểm A:")
xa = float(input("xa: "))
ya = float(input("ya: "))

d = math.sqrt((xa - x0) ** 2 + (ya - y0) ** 2)
if math.isclose(d, R):
    print("Điểm A nằm trên đường tròn")
elif d < R:
    print("Điểm A nằm trong đường tròn")
else:
    print("Điểm A nằm ngoài đường tròn")
```

## Bài 42. Liệt kê các ước của n

```python
n = int(input("Nhap so n = "))
nho, lon = [], []
i = 1
while i * i <= n:            # chỉ duyệt đến √n
    if n % i == 0:
        nho.append(i)
        if i != n // i:
            lon.append(n // i)
    i += 1
for u in nho + lon[::-1]:   # in theo thứ tự tăng dần
    print(u)
```

## Bài 45. Kiểm tra số nguyên tố

## Bài 46. Kiểm tra số hoàn hảo

```python
n = int(input("nhập n: "))

tong = 0
if n > 1:
    tong = 1                 # 1 luôn là ước (không kể chính n)
    i = 2
    while i * i <= n:
        if n % i == 0:
            tong += i
            if i != n // i:
                tong += n // i
        i += 1

if n > 1 and tong == n:
    print(n, "là số hoàn hảo")
else:
    print(n, "không là số hoàn hảo")
```

## Bài 47. Các số hoàn hảo nhỏ hơn n

```python
def la_hoan_hao(x):
    if x < 2:
        return False
    tong = 1
    i = 2
    while i * i <= x:
        if x % i == 0:
            tong += i
            if i != x // i:
                tong += x // i
        i += 1
    return tong == x

n = int(input("Nhap so n: "))
for i in range(1, n):
    if la_hoan_hao(i):
        print(i)
```
