## Bài 83.

```python
import random

def cap_so_cong(A):
    if len(A) <= 2:
        return True
    d = A[1] - A[0]
    for i in range(2, len(A)):
        if A[i] - A[i - 1] != d:
            return False
    return True
n = int(input("Nhập số lượng phần tử của dãy số, N = "))
A = []
for i in range(n):
    A.append(random.randint(-100, 100))
print("Mảng A:")
for x in A:
    print(x, end=" ")
print()
if cap_so_cong(A):
    print("Đây là cấp số cộng")
else:
    print("Đây không là cấp số cộng")
```
